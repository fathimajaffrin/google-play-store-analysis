## 4. Data Type Conversion

```python
import numpy as np
```

```python
def clean_size(value):
    if value == "Varies with device":
        return np.nan
    
    elif value.endswith("M"):
        return float(value.replace("M", ""))
    
    elif value.endswith("k"):
        return float(value.replace("k", "")) / 1024
    
    else:
        return np.nan
```

```python
clean_size("19M")
```

```python
clean_size("8.7M")
```

```python
clean_size("467k")
```

```python
clean_size("Varies with device")
```

## Key Insights

1. Most apps in the dataset have relatively high ratings, with an average rating of approximately 4.17.

2. Paid apps have a slightly higher average rating than free apps.

3. Price has very little relationship with app rating.

4. Installs and ratings have almost no linear relationship.

5. Installs and reviews have a very strong positive relationship after log transformation.

6. GAME has the highest total number of installs, while COMMUNICATION has the highest average installs per app.

7. App size has only a weak relationship with installs.

8. Most apps were updated in the later years of the dataset, particularly 2017 and 2018.

## Conclusion

The Google Play Store dataset shows that app popularity and app ratings are not strongly connected. Although highly installed apps tend to receive substantially more reviews, the number of installs has almost no relationship with rating.

Paid apps have a slightly higher average rating than free apps, but price itself has little relationship with rating. This suggests that simply charging more does not necessarily result in better user ratings.

Among categories, GAME has the highest total number of installs, while COMMUNICATION has the highest average installs per app.

Overall, the analysis highlights the importance of considering multiple metrics when evaluating app performance rather than relying on ratings, installs, or price alone.
